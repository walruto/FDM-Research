```python
import pandas as pd, hashlib, shutil
from pathlib import Path

src = Path("/content/malimg/malimg_paper_dataset_imgs")    # your copy: <family>/<file>.png
exp = Path("/content/fdm/x/FDM-experiments")
dst = exp / "raw_data" / "malimg_dataset"                   # where their config.py looks
idx = pd.read_csv(exp / "data" / "malimg_index.csv")        # 8,444 rows: path, family, md5, label, fold

ok, missing, bad = 0, [], []
for r in idx.itertuples():
    f = src / r.family / r.path.split("/")[-1]              # find it by family + filename
    if not f.exists():
        missing.append(r.path); continue
    if hashlib.md5(f.read_bytes()).hexdigest() != r.md5:    # exact same bytes as the authors' file?
        bad.append(r.path); continue
    out = dst / r.path                                      # e.g. train/Adialer.C/xxx.png
    out.parent.mkdir(parents=True, exist_ok=True)
    shutil.copy2(f, out)
    ok += 1

print("copied:", ok, "| missing:", len(missing), "| md5 mismatch:", len(bad))
print(missing[:5], bad[:5])
```

```python
import json
exp = "/content/fdm/x/FDM-experiments"
load = lambda p: {json.loads(l)["run_id"]: json.loads(l) for l in open(p)}
mine   = load(f"{exp}/results/e3_autoencoder/runs.jsonl")          # your run
theirs = load(f"{exp}/results_authors/e3_autoencoder/runs.jsonl")  # their log

mb, tb = mine["cnn_baseline|fold0"], theirs["cnn_baseline|fold0"]  # baselines, for time ratios
print(f"{'run':22s} {'my acc':>7s} {'their':>7s} {'diff':>6s} | {'my s':>6s} {'their s':>7s} | {'my x':>5s} {'their x':>7s}")
for rid, m in mine.items():
    t = theirs[rid]
    print(f"{rid:22s} {100*m['acc']:7.2f} {100*t['acc']:7.2f} {100*(m['acc']-t['acc']):+6.2f} | "
          f"{m['total_train_seconds']:6.0f} {t['total_train_seconds']:7.0f} | "
          f"{m['total_train_seconds']/mb['total_train_seconds']:5.2f} {t['total_train_seconds']/tb['total_train_seconds']:7.2f}")
```

```python
print(f"{'run':22s} {'epochs m/t':>11s} | {'AE pre s m/t':>13s} | {'clf s m/t':>11s} | {'s/epoch m/t':>12s}")
for rid, m in mine.items():
    t = theirs[rid]
    print(f"{rid:22s} {m['epochs_run']:>4d}/{t['epochs_run']:<5d} | {m['ae_pretrain_seconds']:5.0f}/{t['ae_pretrain_seconds']:<6.0f} | "
          f"{m['clf_train_seconds']:4.0f}/{t['clf_train_seconds']:<5.0f} | {m['seconds_per_epoch']:5.2f}/{t['seconds_per_epoch']:<5.2f}")
```

```python
import json
exp = "/content/fdm/x/FDM-experiments"
load = lambda p: {json.loads(l)["run_id"]: json.loads(l) for l in open(p)}
mine   = load(f"{exp}/results/e3_autoencoder/runs.jsonl")
theirs = load(f"{exp}/results_authors/e3_autoencoder/runs.jsonl")

within = [rid for rid in mine if abs(mine[rid]["acc"] - theirs[rid]["acc"]) <= 0.01]   # 0.01 = 1 point
folds  = sorted({m["fold"] for m in mine.values()})
print(f"folds done: {folds}")
print(f"within 1 point: {len(within)} of {len(mine)}")
print("over 1 point:", [(rid, round(100*(mine[rid]['acc']-theirs[rid]['acc']), 2)) for rid in mine if rid not in within])
```
